```python
ranking = ranking.drop(columns=["RETURNTOPLAY"])
ranking.columns
```

```python
ranking["STANDINGSDATE"] = pd.to_datetime(ranking["STANDINGSDATE"])

ranking["STANDINGSDATE"].dtype
```

```python
ranking.columns = ranking.columns.str.lower()
```

```python
from sqlalchemy import create_engine
engine = create_engine(
    "postgresql+psycopg2://postgres:"password"@localhost:"portnumber"/nba_analysis"
)
ranking.to_sql(
    "rankings",
    engine,
    if_exists="replace",   # use replace ONLY if table doesn't exist
    index=False
)
```
