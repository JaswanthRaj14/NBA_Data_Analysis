```python
players = players.drop_duplicates(subset=["PLAYER_ID", "SEASON"], keep="first")
players.duplicated(subset=["PLAYER_ID", "SEASON"]).sum()
```

```python
players = players[["PLAYER_ID", "PLAYER_NAME", "SEASON"]]
players.head()
```

```python
players = players.rename(columns={
    "PLAYER_ID": "player_id",
    "PLAYER_NAME": "player_name",
    "SEASON": "season"
})
```

```python
from sqlalchemy import create_engine
engine = create_engine("postgresql+psycopg2://postgres:"password"@localhost:"portnumber"/nba_analysis")
players.to_sql("players", engine, if_exists="replace",index=False)
```
