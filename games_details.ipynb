```python
import pandas as pd
```

```python
games_details = pd.read_csv("games_details.csv")
```

```python
games_details["NICKNAME"] = games_details["NICKNAME"].astype("string")
games_details["NICKNAME"].dtype
```

```python
games_details.describe()
```

```python
games_details = games_details.drop(columns=["NICKNAME"])
```

```python
games_details.isnull().sum()
```

```python
games_details["START_POSITION"] = games_details["START_POSITION"].fillna("Bench")
games_details["START_POSITION"].value_counts().head()
```

```python
games_details["COMMENT"] = games_details["COMMENT"].fillna("Played")
games_details["COMMENT"].value_counts().head()
```

```python
games_details["COMMENT"] = (
    games_details["COMMENT"]
    .str.strip()
    .replace({
        "DND - Injury/Illness": "DNP - Injury/Illness"
    })
)
games_details["COMMENT"].value_counts()
```

```python
def normalize_comment(val):
    if val == "Played":
        return "Played"
    elif "Coach" in val:
        return "DNP - Coach's Decision"
    else:
        return "DNP - Injury/Illness"

games_details["COMMENT"] = games_details["COMMENT"].apply(normalize_comment)
games_details["COMMENT"].value_counts()
```

```python
stat_cols = [
    "MIN", "FGM", "FGA", "FG_PCT", "FG3M", "FG3A", "FG3_PCT",
    "FTM", "FTA", "FT_PCT", "OREB", "DREB", "REB",
    "AST", "STL", "BLK", "TO", "PF", "PTS", "PLUS_MINUS"
]
```

```python
games_details[stat_cols] = games_details[stat_cols].apply(
    pd.to_numeric, errors="coerce"
)
```

```python
games_details[stat_cols] = games_details[stat_cols].fillna(0)
games_details[stat_cols].isnull().sum()
```
