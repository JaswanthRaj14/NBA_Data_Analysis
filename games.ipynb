```python
import pandas as pd
import numpy as np
```

```python
games = pd.read_csv("games.csv")
print(games)
```

```python
games.describe()
```

```python
games.shape # shows no of rows and cols in dataset
```

```python
games.isnull().sum()  # count of the null values in each column in dataset
```

```python
games[games.isnull().any(axis=1)]["GAME_STATUS_TEXT"].value_counts() #counts the missing row values in game_status_text col
```

```python
games = games.dropna(subset=["PTS_home", "PTS_away"])
games.isnull().sum()
```

```python
games.duplicated(subset=["GAME_ID"]).sum()
```

```python
games[games.duplicated(subset=["GAME_ID"], keep=False)].sort_values("GAME_ID").head(10)
```

```python
games = games.drop_duplicates(subset=["GAME_ID"], keep="first")
games.duplicated(subset=["GAME_ID"]).sum()

```

```python
games["GAME_DATE_EST"].dtype
```

```python
games["GAME_DATE_EST"].head()
```

```python
games.describe()
```

```python
games["GAME_DATE_EST"] = pd.to_datetime(games["GAME_DATE_EST"])
games["GAME_DATE_EST"].dtype

```

```python
(games["HOME_TEAM_ID"] == games["TEAM_ID_home"]).all()
(games["VISITOR_TEAM_ID"] == games["TEAM_ID_away"]).all()

```

```python
games["TOTAL_POINTS"] = games["PTS_home"] + games["PTS_away"]
games["WIN_TYPE"] = np.where(
    games["HOME_TEAM_WINS"] == 1,
    "Home Win",
    "Away Win"
)


games = games.rename(columns={
    "GAME_ID": "game_id",
    "GAME_DATE_EST": "game_date",
    "HOME_TEAM_ID": "home_team_id",
    "VISITOR_TEAM_ID": "away_team_id",
    "PTS_home": "pts_home",
    "PTS_away": "pts_away",
    "HOME_TEAM_WINS": "home_team_wins",
    "TOTAL_POINTS": "total_points",
    "WIN_TYPE": "win_type",
    "SEASON": "season"
})
```

```python
from sqlalchemy import create_engine
engine = create_engine(
    "postgresql+psycopg2://postgres:"password"@localhost:"portnumber"/nba_analysis"
)
```

```python
games.to_sql(
    "games",
    engine,
    if_exists="replace",
    index=False
)
```
