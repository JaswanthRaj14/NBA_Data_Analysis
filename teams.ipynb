```python
teams = teams[
    ["TEAM_ID", "ABBREVIATION", "NICKNAME", "CITY", "YEARFOUNDED"]
]
teams.head()
```

```python
teams["CITY"] = teams["CITY"].str.title()
teams["NICKNAME"] = teams["NICKNAME"].str.title()
teams["ABBREVIATION"] = teams["ABBREVIATION"].str.upper()
teams.head()
```

```python
teams = teams.rename(columns={
    "TEAM_ID": "team_id",
    "ABBREVIATION": "abbreviation",
    "NICKNAME": "team_name",
    "CITY": "city",
    "YEARFOUNDED": "year_founded"
})
teams.columns
```

```python
from sqlalchemy import create_engine
engine = create_engine(
    "postgresql+psycopg2://postgres:"password"@localhost:"portnumber"/nba_analysis"
)
teams.to_sql(
    "teams",
    engine,
    if_exists="replace",   # use replace ONLY if table doesn't exist
    index=False
)
```
